### **Objective**
Using dictionaries, loops, functions, and conditionals, create a Python program that manages student
grades using a dictionary. The dictionary should store student names as keys and grades as values. Your
program must provide a menu-driven system that allows the user to:
- Add a new student's grade
- View all grades
- View a specific student’s grade
- Update a grade
- Delete a student’s grade
- Exit the program

```python
dict_1 = {"Ali" : 85, "Amna" : 78, "Usman" : 98, "Rabia" : 45}
print(dict_1)
```

### **1. add_grade()**
Purpose:
- Add a new student and their grade to the dictionary.

Requirements:
- Ask user to enter a student’s name
- Ask for the grade
- Add the pair to the dictionary
- Show a confirmation message

```python
# Asking user to enter student name and grade
name = input("Enter student name:")
print("Enter student name:", name)

grade = input("Enter grade:")
print("Enter grade:", grade)

# Adding pair to dictionary
dict_1[name] = grade
print(dict_1)

#Confirmation message
print(f"{name}'s grade added successfully!")
```

### **2. show_all_grades()**
Purpose:
- Display all students and their grades.

Requirements:
- If dictionary is empty, show appropriate message
- Otherwise, print each name and grade

```python
if dict_1 == {}:
    print("No grades available")
    
else:
    for name, grade in dict_1.items():
        print(name, ":", grade)
```

### **3. show_specific_grade()**
Purpose:
- Show the grade for a specific student.

Requirements:
- Ask for student name
- If exists, print grade
- If not, print “Student not found.”

```python
# Asking for student name
name_1 = input("Enter student name:")
print("Enter student name:", name_1)

if name_1 in dict_1:
    grade = dict_1[name_1]
    print(f"{name_1}'s Grade: {grade}")
    
else:
    print("Student not found") 
```

### **4. update_grade()**
Purpose:
- Update the grade of an existing student.

Requirements:
- Ask for student name
- If exists, ask for new grade
- Update the dictionary
- Otherwise, show error message

```python
# Asking for the student name
name_2 = input("Enter student name:")
print("Enter student name:", name_2)

if name_2 in dict_1:
    grade_1 = input("Enter new grade:")
    dict_1[name_2] = grade_1
    print("Enter new grade:", grade_1) 
    print(f"{name_2}'s grade updated successfully!")
    
else:
    print("Student not found.")
    
# Updated dictionary
print("\nUpdated dictionay:", dict_1)
```

### **5. delete_grade()**
Purpose:
- Delete a student's grade record.

Requirements:
- Ask for student name
- If exists, delete entry
- Otherwise show error message


```python
name_3 = input("Enter student name to delete:")
print("Enter student name to delete:", name_3)

if name_3 in dict_1:
    del dict_1[name_3]
    print(f"{name_3}'s record deleted successfully!")
else:
    print("Student not found.")
    
#Updated dictionary
print("\nUpdated dictionay:", dict_1)
```

### **6. menu()**
Purpose:
- Provide a continuous menu for the user.

Requirements:
- Show options 1–6
- Call correct function based on choice
- Loop until “Exit” is selected

```python
dict_1 = {"Ali" : 85, "Amna" : 78, "Usman" : 98, "Rabia" : 45}
print(dict_1)

def menu():
    
    while True:    
        print("\nGrade Management System\n1. Add Student Grade\n2. Show All Grades\n3. Show Specific Student's Grade\n4. Update Grade\n5. Delete Grade\n6. Exit") 
            
        options = int(input("Choose an option (1-6):"))
        print("\nChoose an option (1-6):", options)
            
        #Add student grade
        if options == 1:
                
            print("---------------ADD STUDENT RECORD---------------")  
                
            name_1 = input("Enter student name:")
            print("Enter student name:", name_1)
                
            grade_1 = input("Enter grade:")
            print("Enter grade:", grade_1)
                
            print(f"{name_1}'s grade added successfully!")
                
            dict_1[name_1] = grade_1

            print("\nUpdated Dictionary:", dict_1)
            
            
        #Shows all grades
        elif options == 2:   
                
            print("---------------All GRADES---------------")     
                
            if dict_1 == {}:
                print("No grades available")
                    
            else:
                for name_1, grade_1 in dict_1.items():
                    print(name_1, ":", grade_1)
            

        #Specific student's grade
        elif options == 3:
                
            print("---------------SHOW SPECIFIC STUDENT'S GRADE---------------")    
                
            name_2 = input("Enter student name:")
            print("Enter student name:", name_2)

            if name_2 in dict_1:
                grade_2 = dict_1[name_2]
                print(f"{name_2}'s grade: {grade_2}")
                
            else:
                print("Student not found.")
                    
        #Updates record
        elif options == 4:
                
            print("---------------UPDATE RECORD---------------")
                
            name_3 = input("Enter student name:")
            print("Enter student name:", name_3)
                
            if name_3 in dict_1:
                grade_3 = input("Enter new grade:")
                dict_1[name_3] = grade_3
                print("Enter new grade:", grade_3)    

                print(f"{name_3}'s grade updated successfully!")
                    
            else:
                print("Student not found.")

            print("Updated Dictionary:", dict_1)
                
        #Deletes record
        elif options == 5:
                
            print("---------------DELETE RECORD---------------")
                
            name_4 = input("Enter student name to delete:")
            print("Enter student name to delete:", name_4)
                
            if name_4 in dict_1:
                del dict_1[name_4]
                    
                print(f"{name_4}'s record deleted successfully!")

            else:
                print("Student not found.")
                    
            print("Updated Dictionary:", dict_1)
                
        #Exits program    
        elif options == 6:
            
            print("---------------EXIT---------------")
            
            print("Exit Program.")    
            break
        
        else:
            print("Invalid option. Choose between 1-6.")
            
menu()
```
